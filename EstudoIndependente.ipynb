```python
def calcular_maior(notas):
    return max(notas)


def calcular_menor(notas):
    return min(notas)


def calcular_media(notas):
    return sum(notas) / len(notas)


def contar_aprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota >= 6.0:
            quantidade += 1

    return quantidade


def contar_reprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota < 6.0:
            quantidade += 1

    return quantidade


notas = []

quantidade = int(input("Digite a quantidade de estudantes: "))

for i in range(quantidade):
    nota = float(input(f"Digite a nota do estudante {i + 1}: "))
    notas.append(nota)

print(f"Maior nota: {calcular_maior(notas):.2f}")
print(f"Menor nota: {calcular_menor(notas):.2f}")
print(f"Média da turma: {calcular_media(notas):.2f}")
print(f"Aprovados: {contar_aprovados(notas)}")
print(f"Reprovados: {contar_reprovados(notas)}")
```

```python
def cadastrar_produtos():
    produtos = {}

    quantidade = int(input("Digite a quantidade de produtos: "))

    for i in range(quantidade):
        print(f"\n--- Produto {i + 1} ---")

        codigo = input("Digite o código: ")
        nome = input("Digite o nome: ")
        preco = float(input("Digite o preço: R$ "))
        estoque = int(input("Digite a quantidade em estoque: "))

        produtos[codigo] = {
            "nome": nome,
            "preco": preco,
            "estoque": estoque
        }

    return produtos


def consultar_produto(produtos, codigo):
    if codigo in produtos:
        produto = produtos[codigo]

        print("\n--- Produto encontrado ---")
        print(f"Código: {codigo}")
        print(f"Nome: {produto['nome']}")
        print(f"Preço: R$ {produto['preco']:.2f}")
        print(f"Estoque: {produto['estoque']}")
    else:
        print("Produto não encontrado.")


def calcular_valor_estoque(produtos):
    total = 0

    for produto in produtos.values():
        total += produto["preco"] * produto["estoque"]

    return total


def produto_maior_valor(produtos):
    maior = None

    for codigo, produto in produtos.items():
        if maior is None or produto["preco"] > maior["preco"]:
            maior = {
                "codigo": codigo,
                "nome": produto["nome"],
                "preco": produto["preco"]
            }

    return maior


produtos = cadastrar_produtos()

print("\n--- Consulta de produto ---")
codigo = input("Digite o código do produto que deseja consultar: ")
consultar_produto(produtos, codigo)

total = calcular_valor_estoque(produtos)

print("\n--- Estoque ---")
print(f"Valor total do estoque: R$ {total:.2f}")

maior = produto_maior_valor(produtos)

print("\n--- Produto de maior valor unitário ---")
print(f"Código: {maior['codigo']}")
print(f"Nome: {maior['nome']}")
print(f"Preço: R$ {maior['preco']:.2f}")
```
