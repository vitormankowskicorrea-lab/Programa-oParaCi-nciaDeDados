```python
NomeAluno = str(input("Informe o nome do aluno: "))
x1 = int(input("Informe o valor da nota: "))
x2 = int(input("Informe o valor da nota: "))
x3 = int(input("Informe o valor da nota: "))

x1float = float(x1)
x2float = float(x2)
x3float = float(x3)

mediaNota = (x1float + x2float + x3float )/3
print (f"o aluno {NomeAluno} tirou as notas: {x1float, x2float, x3float} e a media é: {mediaNota}" )
```
